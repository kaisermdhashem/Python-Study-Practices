## Dictionary

```python
recipe = {"pasta": 500, "tomatoes": 400, "garlic": 20, "basil": 10, "olive_oil": 30, "salt": 5}

print(recipe["pasta"])
print(recipe.values())
print(recipe.keys())
print(recipe)
print(recipe.items())

recipe["parmesan"] = 50
recipe["pasta"] = 1000
# Duplicate keys are not allowed/accepted. It updates previous value
```

## Set and Tuple

```python
ingredients = {"pasta", "tomatoes", "pasta", "basil", "garlic", "olive_oil", "salt"}
print(ingredients)

ingredients_list = ["pasta", "tomatoes", "basil", "garlic", "olive_oil", "pasta", "salt"]
unique_ingredients = set(ingredients_list)
print(type(unique_ingredients))
print(unique_ingredients)

ingredients = {"pasta", "tomatoes", "pasta", "basil", "garlic", "olive_oil", "salt"}
sorted_ingredients = sorted(ingredients)
print(type(sorted_ingredients))
print(sorted_ingredients)

serving_sizes = (1, 2, 4, 6, 8)
print(type(serving_sizes))
print(serving_sizes)
print(serving_sizes[1])

ingredients_tuple = tuple(ingredients_list)
print(type(ingredients_tuple))
print(ingredients_tuple)
```

## Conditional Statements & Operators
Comparison operators: ==, !=, >, <, >=, <=

Conditional statements: if, elif, else

Booleans for comparing variables, data structures and values



```python
# Boolean variable
the_truth = True
print(the_truth)

# Compare if 2 is equal to 3
print(2 == 3)

# Compare if 2 is not equal to 3
print(2 != 3)

# Is 5 less than 7?
print(5 < 7)

# Is 5 less than or equal to 7?
print(5 <= 7)

# Is 5 greater than 7?
print(5 > 7)

# Is 5 greater than or equal to 7?
print(5 >= 7)

# Is Ahmed greater than Mohammad
print("Ahmed" > "Mohammad") # Strings ('A', 'M') are evaluated in alphabetic order

# Check pasta quantities
required_quantity = 500
pasta_quantity = 200

# Compare pasta quatities
if pasta_quantity >= required_quantity:
  print("You have enough pasta!")
elif pasta_quantity >= 300:
  print("Nealy enough pasta. Try a smaller portion.")
else:
  print("Not enough pasta.")

ingredients_needed = {"tomatoes": 1000}
pantry_stock = {"tomatoes": 500}
# Check if you have enough tomatoes for the full party
if pantry_stock["tomatoes"] >= ingredients_needed["tomatoes"]:
    print("Enough tomatoes for the party!")
# Check if you have enough for a smaller gathering
elif pantry_stock["tomatoes"] >= 800:
    print("Only enough tomatoes for a smaller gathering.")
# Handle the case where you need to buy more
else:
    print("Need to buy tomatoes before the party.")

basil_grams = 100
required_basil = 200
# Check if you have exactly the right amount of basil
if basil_grams == required_basil:
    print('Perfect! You have exactly the right amount of basil.')
else:
    print('You need to adjust your basil quantity.')
```
